### Enumerate

`enumerate` is another way to loop through lists if you need to access the values and indices. `enumerate` outputs two things for every item in your list, both the index and its corresponding list entry. Check it out below. 

```python
for item in enumerate(todo) : # marks down the list numerical value 
    print(item)
```

Since `enumerate` evaluates to (`index`,`item`) pairs, you can ask the `for` loop for both of those separately. 

```python
for index,entry in enumerate(todo) :
    print(index,entry) # takes away paranthases on out side words along with commas and assigns them numerical value 
```

```python
hours = 45 #initalizing hours and pay 
rate = 10

if hours > 40: #check if total hours exceed standard 40 
    overtime_hours = hours - 40 # creating an if statment for overtime 
    pay = (40 * rate) + (overtime_hours * rate * 1.5) # calculating how many hours count as overtime 
else:
    pay= hours * rate# calculate regular pay without overtime 
pay # display final 
```

### Let's get a little weird: Mutating, Copies, and when two lists are the same 

I think it is more natural to show you this in the opposite order of the book. There are all sorts of reasons why you might want to copy a list. Most of those reasons include changing items on the new list without changing the old list. Unfortunately, this doesn't quite work how you might like it to. See my example below - I expect you may have questions, please ask! 

```python
old_list = ['yeller','faithful','navy','testament','bay','school']
new_list = old_list

new_list, old_list
```

```python
# "is" is how you check if two objects in python are the same.

new_list is old_list
```

```python
# Okay great, let's try to add a new item that is more new. 

new_list.append('delhi')# adding it to the end of the list 
new_list
```

```python
old_list

#Uh-Oh! Delhi was also added to the old list :( 
```

When you use assignment (`=`) to make a copy of a list, you actually aren't making a copy of it. You are just giving the same list a second name. It would be like if you made a website, and then had two different links to that website. The links are going to the same place, so changing the website at one of the links would also affect the other. Assignment in Python works exactly the same way. 

Keep going for more info and how to do it properly. 

```python
# Let's reset our list
old_list = ['yeller','faithful','navy','testament','bay','school']

# One not very exciting way to do it is to just write every element in the list again. 
new_list = ['yeller','faithful','navy','testament','bay','school']

# We use "is" to verify that these two lists, despite having the same elements, are different according to python
old_list is new_list
```

```python
new_list.append('delhi')
```

```python
# It worked! 

old_list,new_list
```

```python
# The best way to do it is with slicing!

new_list = old_list[:]

# Check whether or not they are the same object.
new_list is old_list
```

```python
new_list.remove('faithful')

new_list, old_list
```

```python
old_list = ['baker','faithful','blue','heard','sunny','school']

 
new_list = ['baker','faithful','blue','heard','sunny','school']


old_list is new_list


new_list.append('honey bun')

old_list, new_list

new_list = old_list[:]

new_list is old_list

new_list.remove('baker')

new_list, old_list
```
